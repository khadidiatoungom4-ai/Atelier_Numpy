# Partie 4 – Indexation et slicing 

Considérant le tableau représentant les 24 températures d'une journée, afficher : 
1) la première mesure ;  

```python
import numpy as np
np.random.seed(42)
temperatures = np.random.uniform(20, 35, 24)
print("Températures :")
print(temperatures)
```

2) la dernière mesure ;  

```python
print("Dernière mesure :", temperatures[-1])
```

3) la température de 8h ; 

```python
print("Température à 8h :", temperatures[8])
```

4) les températures de 8h à 12h ;  

```python
print("Températures de 8h à 12h :", temperatures[8:13])
```

5) les températures des 6 premières heures ;  

```python
print("Températures des 6 premières heures :", temperatures[:6])
```

6) les températures des heures paires. 

```python
print("Températures des heures paires :", temperatures[0:24:2])
```
