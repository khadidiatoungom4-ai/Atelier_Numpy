# Partie 5 – Filtrage booléen 

Considérant le tableau représentant les 24 températures d'une journée, on souhaite détecter les 
situations inhabituelles.  
1) Afficher les températures supérieures à 30 °C 

```python
import numpy as np
np.random.seed(42)
temperatures = np.random.uniform(20, 35, 24)
print("Températures :")
print(temperatures)
```

```python
temperatures_sup_30 = temperatures[temperatures > 30]
print("Températures supérieures à 30 °C :")
print(temperatures_sup_30)
```

2) Afficher les températures « normales » càd comprises entre 20 °C et 30 °C

```python
temperatures_normales = temperatures[
    (temperatures >= 20) & (temperatures <= 30)
]
print("Températures normales :")
print(temperatures_normales)
```

# Partie 7 – Concaténation 

On suppose qu’on a un 2nd bâtiment avec 3 capteurs.  
1) Créer aléatoirement la matrice représentant les températures mesurées par les 3 capteurs

```python
import numpy as np
np.random.seed(100)
temperatures_batiment2 = np.random.uniform(20, 35, (3, 24))
print("Matrice du 2ème bâtiment :")
print(temperatures_batiment2)
print("Forme :", temperatures_batiment2.shape)
```

2) Déterminer de deux manières différentes la matrice qui fusionne les températures mesurées par 
les 8 capteurs durant une journée 


-      Premiere methode

```python
temperatures_8_capteurs = np.concatenate(
    (temperatures, temperatures_batiment2),
    axis=0
)
print("Matrice des 8 capteurs :")
print(temperatures_8_capteurs)
print("Forme :", temperatures_8_capteurs.shape)
```

-          Deuxieme methode

```python
temperatures_8_capteurs_v2 = np.vstack(
    (temperatures, temperatures_batiment2)
)
print("Matrice des 8 capteurs :")
print(temperatures_8_capteurs_v2)
print("Forme :", temperatures_8_capteurs_v2.shape)
```
